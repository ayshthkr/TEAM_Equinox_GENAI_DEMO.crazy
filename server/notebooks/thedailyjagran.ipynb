```python
import urllib.request
from urllib.error import URLError, HTTPError
from bs4 import BeautifulSoup
from urllib.parse import urljoin

url = "https://www.thedailyjagran.com/india"   
cms_links = set()
def thedailyjagran(url):
    

    try:
        print("Attempting:", url)

        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                          "AppleWebKit/537.36 (KHTML, like Gecko) "
                          "Chrome/115.0 Safari/537.36"
        }
        req = urllib.request.Request(url, headers=headers)

        with urllib.request.urlopen(req) as response:
            if response.status == 200 and "text/html" in response.getheader("Content-Type", ""):
                html = response.read().decode("utf-8")
                soup = BeautifulSoup(html, "html.parser")

                # ✅ get ALL <a> tags no matter where nested
                for a in soup.find_all("a", href=True):
                    href = a["href"].strip()

                    # convert relative → absolute
                    full_url = urljoin(url, href)

                    # ✅ only keep links from /india section
                    if "thedailyjagran.com/india" in full_url:
                        cms_links.add(full_url)

                print(f"Found {len(cms_links)} India links:\n")
                for link in cms_links:
                    print(link)

            else:
                print("Not scrapable: content type is not HTML or bad status")

    except HTTPError as e:
        print(f"HTTP Error: {e.code}")
    except URLError as e:
        print(f"URL Error: {e.reason}")

    return cms_links


# 🔥 run it
links = thedailyjagran(url)
```

```python
import urllib.request
from urllib.error import URLError, HTTPError
from bs4 import BeautifulSoup

def contentoftoi(url):
    print(url)

    try:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                          "AppleWebKit/537.36 (KHTML, like Gecko) "
                          "Chrome/115.0 Safari/537.36"
        }
        req = urllib.request.Request(url, headers=headers)

        with urllib.request.urlopen(req) as response:
            if response.status == 200 and "text/html" in response.getheader("Content-Type", ""):
                html = response.read().decode("utf-8")
                soup = BeautifulSoup(html, "html.parser")
                
               
                # ✅ Step 1: find outer container
                outer = soup.find("div", class_="container px-6 lg:px-0 pt-6 lg:pt-6 pb-6 lg:pb-10")

                if outer:
                    # ✅ Step 2: find nested div inside outer
                    target = outer.find("div", class_="flex justify-between flex-wrap")

                    if target:
                        # ---- Extract text ----
                        full_text = target.get_text(separator="\n", strip=True)
                        lines = full_text.split("\n")
                        article_lines = []

                        for line in lines:
                            if line.strip().lower().startswith("end of article"):
                                break
                            article_lines.append(line)

                        article_text = "\n".join(article_lines)

                        # ---- Extract images ----
                        img_urls = []
                        for img in target.find_all("img"):
                            if img.has_attr("src"):
                                img_urls.append(img["src"])
                            elif img.has_attr("data-src"):
                                img_urls.append(img["data-src"])

                        # ---- Print result ----
                        print("=== Article Text ===")
                        print(article_text)
                        print("\n=== Image URLs ===")
                        for u in img_urls:
                            print(u)

                    else:
                        print("Inner flex-wrap container not found")
                else:
                   return "Outer container not found"

            else:
                print("Not scrapable: content type is not HTML or bad status")

    except HTTPError as e:
        print(f"HTTP Error: {e.code}")
    except URLError as e:
        print(f"URL Error: {e.reason}")
```

```python
cms_links = list(cms_links)
cnt=0
for i in cms_links:

    a=contentoftoi(i)
    if a=="Outer container not found":
        cnt+=1
```
