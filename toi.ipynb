```python
import urllib.request
from urllib.error import URLError, HTTPError
from bs4 import BeautifulSoup

url = "https://timesofindia.indiatimes.com/india"

try:
    # fetch page
    with urllib.request.urlopen(url) as response:
        if response.status == 200 and "text/html" in response.getheader("Content-Type", ""):
            html = response.read().decode("utf-8")
            soup = BeautifulSoup(html, "html.parser")

            # get all <a> tags with href ending in .cms
            

            cms_links = []
            for a in soup.find_all("a", href=True):
                if a["href"].endswith(".cms") and a['href'].startswith("https://timesofindia.indiatimes.com/india"):
                    cms_links.append(a["href"])

            print(f"Found {len(cms_links)} links ending with .cms:\n")
            for link in cms_links:
                print(link)
                print(fetch(link))
        else:
            print("Not scrapable: content type is not HTML or bad status")
except HTTPError as e:
    print(f"HTTP Error: {e.code}")
except URLError as e:
    print(f"URL Error: {e.reason}")
```

```python
import urllib.request
from urllib.error import URLError, HTTPError
from bs4 import BeautifulSoup

url = "https://timesofindia.indiatimes.com/india"

try:
    with urllib.request.urlopen(url) as response:
        if response.status == 200 and "text/html" in response.getheader("Content-Type", ""):
            html = response.read().decode("utf-8")
            soup = BeautifulSoup(html, "html.parser")

            cms_links = []

            # iterate all a + span in DOM order
            for tag in soup.find_all(["a", "span"]):
                if tag.name == "a" and tag.has_attr("href"):
                    if tag["href"].endswith(".cms") and tag["href"].startswith("https://timesofindia.indiatimes.com/india"):
                        cms_links.append(tag["href"])

                # stop if span class has id-r-component AND text is "end of article"
                if (
                    tag.name == "span"
                    and "id-r-component" in tag.get("class", [])
                    and tag.get_text(strip=True).lower() == "end of article"
                ):
                    break

            print(f"Found {len(cms_links)} links ending with .cms (before 'end of article'):\n")
            for link in cms_links:
                print(link)
        else:
            print("Not scrapable: content type is not HTML or bad status")
except HTTPError as e:
    print(f"HTTP Error: {e.code}")
except URLError as e:
    print(f"URL Error: {e.reason}")
```

```python
import urllib.request
from urllib.error import URLError, HTTPError
from bs4 import BeautifulSoup

def contentoftoi(url):
    print(url)

    try:
        with urllib.request.urlopen(url) as response:
            if response.status == 200 and "text/html" in response.getheader("Content-Type", ""):
                html = response.read().decode("utf-8")
                soup = BeautifulSoup(html, "html.parser")

                body = soup.body
                # if class wiht undefined xciRh then ignore
                if body:
                    # extract body text
                    bad_classes = ["xciRh", "Mivr2 ZwGeW", "Ec__7 AuiwR undefined","MvyrO  rel","FNwgv FhCmn"]

                    for bad in body.find_all(attrs={"class": lambda c: c and any(b in c for b in bad_classes)}):
                        bad.decompose()
                    full_text = body.get_text(separator="\n", strip=True)

                    # split into lines
                    lines = full_text.split("\n")

                    article_lines = []
                    for line in lines:
                        # stop when "End of Article" is found
                        if line.strip().lower().startswith("end of article"):
                            article_lines.append(line)
                            break
                        article_lines.append(line)

                    # join back cleaned article
                    article_text = "\n".join(article_lines)

                    print(article_text)

                else:
                    print("No <body> tag found")

            else:
                print("Not scrapable: content type is not HTML or bad status")

    except HTTPError as e:
        print(f"HTTP Error: {e.code}")
    except URLError as e:
        print(f"URL Error: {e.reason}")
```
